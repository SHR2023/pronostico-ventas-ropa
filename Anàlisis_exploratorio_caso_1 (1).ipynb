```python
import pandas as pd
import numpy as np
from scipy.stats import skew

# Variables numéricas
num_vars = [
    "edadloc", "correo", "paginas", "telefono",
    "impresa", "servicio", "nomina", "ropamujer"
]

# Crear tabla descriptiva
desc_table = pd.DataFrame({
    "n": train[num_vars].count(),
    "Media": train[num_vars].mean(),
    "Mediana": train[num_vars].median(),
    "Desv_std": train[num_vars].std(),
    "Min": train[num_vars].min(),
    "Max": train[num_vars].max(),
    "Asimetria": train[num_vars].apply(skew)
})

# Redondear
desc_table = desc_table.round(2)

# Mostrar tabla
print(desc_table)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns

# Variables predictoras
X = train.drop(columns=["ropamujer", "idloc"])

# One-hot encoding
X = pd.get_dummies(X, drop_first=True)

# Variable objetivo
y = train["ropamujer"]

# Modelo
rf = RandomForestRegressor(
    n_estimators=300,
    random_state=42
)

rf.fit(X, y)

# Importancia
importance = pd.DataFrame({
    "Variable": X.columns,
    "Importancia": rf.feature_importances_
})

importance = importance.sort_values(
    by="Importancia",
    ascending=False
)

# Top 10
top10 = importance.head(10)

# Gráfico
plt.figure(figsize=(10,6))

sns.barplot(
    data=top10,
    x="Importancia",
    y="Variable"
)

plt.title("Importancia de variables")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Modelo lineal
model = LinearRegression()

model.fit(X, y)

# Predicciones
y_pred = model.predict(X)

# Residuos
residuos = y - y_pred

# Gráfico
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=y_pred,
    y=residuos
)

plt.axhline(
    y=0,
    color='red',
    linestyle='--'
)

plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")

plt.title("Residuos vs Valores ajustados")

plt.show()
```
